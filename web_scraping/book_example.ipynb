Book example

```python
import bs4

import requests

page = 1
page_num = 1

two_stair_products = []

for i in range(1, page_num + 1):
    base_url = 'https://books.toscrape.com/catalogue/page-{}.html'
    result = requests.get(base_url.format(page))

    soup = bs4.BeautifulSoup(result.text, 'lxml')
    products = soup.select('article.product_pod')

    for product in products:
        if len(product.select('.star-rating.Two')) != 0:
            # Title
            title = product.select('h3 a')[0]['title']

            # Rating
            stars = product.select_one('.star-rating')
            rating = stars["class"][1]

            print(f'Product: {title} has {rating} stars')

            two_stair_products.append(title)

    page += 1

print(two_stair_products)
```
