```python
import requests
import bs4

url = 'https://en.wikipedia.org/wiki/Jonas_Salk'
headers = {
    'User-Agent': 'MyResearchScript/1.0 (https://example.org; user@example.com)'
}

result = requests.get(url, headers=headers)

soup = bs4.BeautifulSoup(result.text, 'lxml')

items = soup.select('.vector-toc-text span:nth-child(2n)')
first_item = items[0]
print('First item: ' + first_item.getText())
print('All items: ')
for item in items:
    print(item.getText())

```
