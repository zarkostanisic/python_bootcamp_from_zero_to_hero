```python
import requests
import bs4

url = 'https://en.wikipedia.org/wiki/Jonas_Salk'
headers = {
    'User-Agent': 'MyResearchScript/1.0 (https://example.org; user@example.com)'
}

result = requests.get(url, headers=headers)

type(result)

result.text

soup = bs4.BeautifulSoup(result.text, "lxml")

#print(soup)

print('Title')
title = soup.select('title')[0].getText()
#print(type(title))
print(title)

print('Paragraph')
site_paragraphs = soup.select('p')
#print(type(site_paragraphs[0]))
print(site_paragraphs[1].getText())
```
