```python
import requests
import bs4

url = 'https://en.wikipedia.org/wiki/Jonas_Salk'
headers = {
    'User-Agent': 'MyResearchScript/1.0 (https://example.org; user@example.com)'
}

result = requests.get(url, headers=headers)

soup = bs4.BeautifulSoup(result.text, 'lxml')

items = soup.select('.vector-toc-text span:nth-child(2n)')
first_item = items[0]
print('First item: ' + first_item.getText())
print('All items: ')
for item in items:
    print(item.getText())

```

Images

```python
import requests
import bs4
import os

url = 'https://en.wikipedia.org/wiki/Jonas_Salk'
headers = {
    'User-Agent': 'MyResearchScript/1.0 (https://example.org; user@example.com)'
}

result = requests.get(url, headers=headers)

soup = bs4.BeautifulSoup(result.text, 'lxml')

items = soup.select('.infobox img')
first_item = items[0]

print(first_item.get('src'))

i = 0
for item in items:
    src = item['src']
    img_link = requests.get('https:' + src, headers=headers)

    img_content = img_link.content

    f = open(os.getcwd() + '/images/' + str(i) + '_my_computer_image.jpg', 'wb')
    f.write(img_content)
    f.close()

    i += 1

```
