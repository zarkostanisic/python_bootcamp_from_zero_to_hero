Interact Functionality With GUIs

```python
from ipywidgets import interact, interactive, fixed
import ipywidgets as widgets
from IPython.display import display
```

```python
def func1(x):
    return x**2

def func2(y):
    return y

def func3(z):
    return z

# slider
interact(func1, x=10)
# textbox
interact(func2, y=fixed('Hello World!'))
# checkbox
interact(func3, z=True)
```

With Decorators

```python
@interact(x=True,y=fixed(1.0))
def g(x, y):
    return (x, y)
```

Slider - Min and max value

```python
def func(x):
    return x

interact(func, x=widgets.IntSlider(min=-100, max=100, step=1, value=0))

interact(func, x=(-10.0, 10.0, 0.5))
```

```python
def func(x):
    return x
interact(func, x='hello')
```

Dropdown

```python
def func(x):
    return x

#list
interact(func, x=['hello', 'world'])
#dictionary
interact(func, x={'one': 10, 'two': 20})
```
