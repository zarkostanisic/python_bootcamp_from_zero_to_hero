Widget Basics

Display

```python
import ipywidgets as widgets
from IPython.display import display
```

```python
w = widgets.IntSlider()
display(w)
```

Close

```python
w.close()
```

Get value

```python
w.value
```

Set value

```python
w.value = 50
```

Keys

```python
w.keys
```

Max

```python
w.max = 2000
```

Link

```python
a = widgets.FloatText()
b = widgets.FloatSlider()
display(a, b)
mylink = widgets.jslink((a, 'value'), (b, 'max'))
```

Unlink

```python
mylink.unlink()
```
