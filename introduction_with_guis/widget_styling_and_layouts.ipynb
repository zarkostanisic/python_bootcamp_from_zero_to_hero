Widget Styling And Layouts

Layout

```python
import ipywidgets as widgets
from IPython.display import display

w = widgets.IntSlider()
display(w)
w.layout.margin = 'auto'
w.layout.height = '75px'
```

```python
x = widgets.IntSlider(value=15, description='New slider')
display(x)

x.layout = w.layout
```

Style

```python
widgets.Button(description='Ordinary Button', button_style='')
```

```python
widgets.Button(description='Danger Button', button_style='danger')
```

```python
b1 = widgets.Button(description='Custom Color')
b1.style.button_color = 'lightgreen'

display(b1)
```

```python
#b1.style.keys
```

```python
b2 = widgets.Button(description='NEW')
b2.style = b1.style
display(b2)
```

```python
s1 = widgets.IntSlider(description='MyHandle')
s1.style.handle_color = 'blue'
display(s1)
```
