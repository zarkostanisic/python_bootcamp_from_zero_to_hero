Sets

```python
s = set()

print(s)
```

Add

```python
s.add(1)
s.add(2)

print(s)

s.add(2)

print(s)
```

Clear

```python
s.clear()

print(s)
```

Copy

```python
s = {1, 2, 3}
print(s)
sc = s.copy()
print(sc)

print('------')

s.add(4)
print(s)
print(sc)

```

Difference

```python
s.difference(sc)
```

Difference update

```python
s1 = {1, 2, 3}
s2 = {1, 4, 5}

s1.difference_update(s2)
s1
```

Discard

```python
print(s)

s.discard(12)

print(s)
```

Intersection

```python
s1 = {1, 2, 3}
s2 = {1, 2, 4}

s1.intersection(s2)
s1
```

Intersection update

```python
s1.intersection_update(s2)
s1
```

Is disjoint

```python
s1 = {1, 2}
s2 = {1, 2, 4}
s3 = {5}

print(s1.isdisjoint(s2))
print(s1.isdisjoint(s3))
```

Is subset

```python
s1 = {1, 2}
s2 = {1, 2, 4}

print(s1.issubset(s2))
```

Is upperset

```python
s1 = {1, 2}
s2 = {1, 2, 4}

print(s2.issuperset(s1))
```

Symmetric difference

```python
s1 = {1, 2}
s2 = {1, 2, 4}

s1.symmetric_difference(s2)
```

Union

```python
s1 = {1, 2}
s2 = {1, 2, 4}

s1.union(s2)
```

Update

```python
s1 = {1, 2}
s2 = {1, 2, 4}

s1.update(s2)
s1
```
