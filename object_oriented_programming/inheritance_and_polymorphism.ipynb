Inheritance

```python
class Animal():
    def __init__(self):
        print("ANIMAL CREATED")

    def who_am_i(self):
        print("I am an animal")

    def eat(self):
        print("I am eating")

class Dog(Animal):
    def __init__(self):
        Animal.__init__(self)

        print("DOG CREATED")

    # OVERRIDING
    def who_am_i(self):
        print("I am a dog")

    def eat(self):
            print("I am dog and eating")

    # NOT EXISTS IN Animal CLASS
    def bark(self):
        print('WOOF!')
        

print('--- ANIMAL ---')
myanimal = Animal()
myanimal.who_am_i()
myanimal.eat()

print('--- DOG ---')
mydog = Dog()
mydog.who_am_i()
mydog.eat()
mydog.bark()
```

Polymorphism

```python
class Dog():

    def __init__(self, name):
         self.name = name
    
    def speak(self):
        return self.name + " says woof!"

class Cat():

    def __init__(self, name):
         self.name = name
    
    def speak(self):
        return self.name + " says meow!"

print("--- DOG ---")
mydog = Dog('Tom')
print(mydog.speak())

print("--- CAT ---")
mycat = Cat('Tom')
print(mycat.speak())

for pet_class in [mydog, mycat]:
    print(type(pet_class))
    print(type(pet_class.speak()))

def pet_speak(pet):
    print(pet.speak())

pet_speak(mydog)
pet_speak(mycat)
```

```python
class Animal():
    def __init__(self, name):
        self.name = name

    def speak(self):
        raise NotImplementedError("Subclass must implement this abstract method")

class Dog(Animal):
    def speak(self):
        return self.name + "says woof!"

class Cat(Animal):
    def speak(self):
        return self.name + "says meow!"

mydog = Dog('Fido')
mycat = Cat("isis")

print(mydog.speak())
print(mycat.speak())
```
