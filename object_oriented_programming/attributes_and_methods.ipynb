Attributes and methods

```python
class Circle():
    # CLASS OBJECT ATTRIBUTE
    pi = 3.14

    # CONSTRUCTOR
    def __init__(self, radius=1):
        self.radius = radius
        self.area = radius ** 2 * self.pi

    # METHOD
    def get_circumference(self):
        # Circle.pi == self.pi
        return 2 * self.radius * Circle.pi 


circle = Circle(30)
circle.pi
circle.radius
circle.get_circumference()
print(circle.area)

```
