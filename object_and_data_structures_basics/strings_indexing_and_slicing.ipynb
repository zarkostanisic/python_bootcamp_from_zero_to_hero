Indexing

```python
mystring = 'Hello World'
print(mystring)
print(mystring[0])
print(mystring[8])
print(mystring[-2])
```

Test

```python
from unittest import TestCase, main

x = 'Hello World'[-3]

class Evaluate(TestCase):
    def test_exercise(self):
        global x
        s = x.replace(' ', '')
        if '=' in s:
            x = s.split('=')[1]
        else:
            x = s
        self.assertEqual('r', x, "Sorry, wrong letter")

if __name__ == '__main__':
    main(argv=[''], exit=False, failfast=False)
```

Slicing

```python
mystring = "abcdefghijk"
print(mystring[2])
print(mystring[2:])
print(mystring[:3])
print(mystring[3:6])
print(mystring[1:3])
print(mystring[::3])
print(mystring[2:7:2])
print(mystring[::-1])
```

Test

```python
from unittest import TestCase, main

x = 'tinker'[1:4]

class Evaluate(TestCase):
    def test_exercise(self):
        self.assertEqual('ink', x, "Sorry, that was incorrect")

if __name__ == '__main__':
    main(argv=[''], exit=False, failfast=False)
```
