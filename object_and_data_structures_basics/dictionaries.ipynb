Dictionaries

```python
my_dict = {"key1": "value1", "key2": "value2"}

print(my_dict)

print(my_dict['key1'])
```

```python
prices_lookup = {'apple': 2.39, 'oranges': 1.99, 'milk': 5.80}
print(prices_lookup['apple'])
```

```python
d = {'k1': 123, 'k2': [0, 1, 2], 'k3': {'insideKey': 100}}
print(d['k2'])
print(d['k3'])
print(d['k3']['insideKey'])
```

```python
d = {'key1': ['a', 'b', 'c']}
print(d)

letter = d['key1'][2]
print(letter.upper())

d['key1'][2].upper()

d['key1'][2] = d['key1'][2].upper()
print(d['key1'][2])
```

```python
d = {'k1': 100, 'k2': 200}
print(d)

d['k3'] = 300
print(d)

d['k1'] = 'NEW VALUE'
print(d)

print(d.keys())
print(d.values())
print(d.items())
```

```python
from unittest import TestCase, main
import ast

myvar = ["{'Monday': 19, 'Tuesday': 20}"]

class Evaluate(TestCase):
    def test_exercise(self):
        s = myvar
        if '=' in s:
            x = ast.literal_eval(s.split('=')[1])
        else:
            x = ast.literal_eval(s)
        self.assertTrue(type(x)==dict, "Sorry, this is not a dictionary")
        for k,v in x.items():
            self.assertTrue(type(k)==str, "Sorry, one of your keys is not a string")
            self.assertTrue(type(v)==int, "Sorry, one of your values is not an integer")

if __name__ == '__main__':
    main(argv=[''], exit=False, failfast=False)
```
