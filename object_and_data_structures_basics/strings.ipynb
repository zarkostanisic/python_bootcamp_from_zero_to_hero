Test

```python
import sys
from io import StringIO
from unittest import TestCase, main
from unittest.mock import patch

def student_solution():
    print('"Hello World"')


class Evaluate(TestCase):
    def test_hello_world(self):
        with patch('sys.stdout', new_callable=StringIO) as mock_stdout:
            
            student_solution()  
            
            output = mock_stdout.getvalue()  
            self.assertTrue(output in ["Hello World\n","'Hello World'\n",'"Hello World"\n'], "You must print Hello World")
        
if __name__ == '__main__':
    # argv=[''] provides a dummy program name for argv[0] to prevent the IndexError,
    # while leaving the rest of the arguments completely empty.
    main(argv=[''], exit=False, failfast=False)
```
