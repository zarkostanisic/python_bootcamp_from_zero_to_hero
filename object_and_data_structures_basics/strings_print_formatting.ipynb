Format

```python
first_name = "Jose"
last_name = "Clark"
address = "London"

print("Hello " + first_name)
print("Hello {}".format(first_name))
print("Hello {} {} from {}".format(first_name, last_name, address))
print("Hello {1} {0} from {2}".format(first_name, last_name, address))
print("Hello {f} {l} from {a}".format(f=first_name, l=last_name, a=address))
```

```python
result  = 100/777
print("The result was {r:10.5f}".format(r=result))
print("The result was {r:1.2f}".format(r=result))
```

```python
name = "Hose"
print("Hello, his name is {}".format(name))
print(f"Hello, his name is {name}")

name = "Sam"
age = 3

print(f"{name} is {age} years year old")
```
