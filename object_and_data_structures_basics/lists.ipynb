Lists

```python
my_list = [1, 2, 3]

print(my_list)

my_list = ['STRING', 100, 23.2]

print(my_list)
print(len(my_list))

my_list = ['one', 'two', 'three']
print(my_list[0])
my_list = ['one', 'two', 'three']
print(my_list[1:])

another_list = ['four', 'five']

new_list = my_list + another_list
print(new_list)

new_list[0] = 'ONE ALL CAPS'
print(new_list)

new_list.append('six')
print(new_list)

popped_item = new_list.pop()
print(new_list)
print(popped_item)

popped_item = new_list.pop(0)
print(new_list)
print(popped_item)

string_list = ['a', 'e', 'x', 'b', 'c']
number_list = [1, 3, 4, 8]
print(string_list)
print(number_list)

string_list_sorted = string_list.sort()
print(string_list_sorted)

string_list.sort()
string_list_sorted = string_list
print(string_list_sorted)

number_list.reverse()
print(number_list)
```

```python
from unittest import TestCase, main
import ast

# Wrap the expression in quotes so it is passed as a string literal
v = "[1, 'two', 3.14159, False]"

class Evaluate(TestCase):
    def test_exercise(self):
        s = v
        if '=' in s:
            x = ast.literal_eval(s.split('=')[1])
        else:
            x = ast.literal_eval(s)
        y = [type(n) for n in x]
        self.assertTrue(int in y, "Sorry, you forgot to include an integer")
        self.assertTrue(str in y, "Sorry, you forgot to include a string")
        self.assertTrue(float in y, "Sorry, you forgot to include a float")
        self.assertTrue(bool in y, "Sorry, you forgot to include a boolean")


if __name__ == '__main__':
    main(argv=[''], exit=False, failfast=False)
```
