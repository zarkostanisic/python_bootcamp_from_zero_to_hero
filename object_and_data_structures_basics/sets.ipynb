Test

```python
from unittest import TestCase, main

myset = "set('Mississippi')"

class Evaluate(TestCase):
    def test_exercise(self):
        s = myset
        if '=' in s:
            x = eval(s.split('=')[1])
        else:
            x = eval(s)

        self.assertEqual(set('Mississippi'), x, "Sorry, your solution did not return {'M', 'i', 'p', 's'}")

if __name__ == '__main__':
    main(argv=[''], exit=False, failfast=False)
```
