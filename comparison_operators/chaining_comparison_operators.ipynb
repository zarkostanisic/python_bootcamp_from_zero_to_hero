Comparison

```python
compare = 1 < 2
print(compare)

compare = 2 < 3
print(compare)

compare = 1 < 2 < 3
print(compare)

compare = 1 < 2 > 3
print(compare)

compare = 1 < 2 and 2 > 3
print(compare)

compare = 1 < 2 and 2 < 3
print(compare)

compare = 'h' == 'h' and 2 == 2
print(compare)

compare = 100 == 1 or 2 == 2
print(compare)

compare = 1 == 1
print(compare)

compare = not 1 == 1
print(compare)

compare = not 400 > 5000
print(compare)
```
