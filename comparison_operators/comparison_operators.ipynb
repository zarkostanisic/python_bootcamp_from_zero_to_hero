Comparison

```python
compare = 2 == 2
print(compare)

compare = 2 == 1
print(compare)

compare = 'hello' == 'bye'
print(compare)

compare = 'Bye' == 'bye'
print(compare)

compare = '2' == 2
print(compare)

compare = 2.0 == 2
print(compare)

compare = 3 != 3
print(compare)

compare = 4 != 5
print(compare)

compare = 2 > 1
print(compare)

compare = 1 > 2
print(compare)

compare = 1 < 2
print(compare)

compare = 2 < 5
print(compare)

compare = 2 >= 5
print(compare)

compare = 4 <= 5
print(compare)
```
