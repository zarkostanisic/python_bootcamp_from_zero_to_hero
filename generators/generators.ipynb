```python
def create_cubes(n):
    #result = []

    for x in range(n):
        #result.append(x**3)
        yield x**3

    #return result
create_cubes(10)
list(create_cubes(10))
#for x in create_cubes(10):
    #print(x)
```

```python
def gen_fibon(n):
    a = 1
    b = 1

    #output = []
    for i in range(n):
        #output.append(a)
        yield a
        a,b = b,a+b

    #return output

for number in gen_fibon(10):
    print(number)
```

```python
def simple_gen():
    for x in range(3):
        yield x

for number in simple_gen():
    print(number)
```
