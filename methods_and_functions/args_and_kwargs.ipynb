Args and kwargs

Args

```python
def myfunc(a, b, c=0, d=0):
    return sum((a, b, c, d))

result = myfunc(1, 2, 5, 8)
result
```

```python
def myfunc(*args):
    return sum(args)

result = myfunc(1, 2, 3)
result
```

```python
def myfunc(*args):
    for item in args:
        print(item)

result = myfunc(1, 2, 3)
result
```

Kwargs

```python
def myfunc(**kwargs):
    print(kwargs)
    
    if 'fruit' in kwargs:
        print('My fruit choice is {}'.format(kwargs['fruit']))
    else: 
        print('I did not find any fruit here')

result = myfunc(fruit='apple', veggie = 'lettuce')
result
```

Mix args and kwargs

```python
def myfunc(*args, **kwargs):
    print(args)
    print(kwargs)

    print('I would like {} {}'. format(args[0], kwargs['food']))

result = myfunc(10, 2, fruit='apple', food='eggs', animal='dog')
```
