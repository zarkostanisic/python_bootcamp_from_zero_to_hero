```python
def myfunc(*args):
    out = []

    for num in args:
        if num % 2 == 0:
            out.append(num)

    return out

result = myfunc(1, 2, 3)
result

from unittest import TestCase, main

class Evaluate(TestCase):
    def test_five(self):
        x = myfunc(-2, 4)
        self.assertEqual([-2, 4], x, "This is incorrect")

if __name__ == '__main__':
    main(argv=[''], exit=False, failfast=False)
```

Solution 10

```python
def myfunc(x):
    out = []

    for i in range(len(x)):
        if i % 2 == 0:
            out.append(x[i].lower())
        else:
            out.append(x[i].upper())
    return ''.join(out)

result = myfunc('Zarko')
result

from unittest import TestCase, main

class Evaluate(TestCase):
    def test_large(self):
        x = myfunc('supercalifragilisticexpialidocious')
        self.assertTrue(x in ['sUpErCaLiFrAgIlIsTiCeXpIaLiDoCiOuS','SuPeRcAlIfRaGiLiStIcExPiAlIdOcIoUs'], "This is incorrect")

    def test_small(self):
        y = myfunc('Hi')
        self.assertTrue(y in ['hI','Hi'], "This is incorrect")

if __name__ == '__main__':
    main(argv=[''], exit=False, failfast=False)
```
