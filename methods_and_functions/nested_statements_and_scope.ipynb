```python
# GLOBAL
name = 'THIS IS A GLOBAL STRING'

def greet():
    # ENCLOSING
    #name = 'Sammy'

    def hello():
        # LOCAL
        #name = 'IM A LOCAL'
        print('Hello ' + name)

    hello()

greet()

# BUILT
help(len)
```

```python
x = 50

def func(x):
    print(f'X is {x}')

    # LOCAL REASSIGNMENT!
    x = 200

    print(f'I JUST LOCALLY CHANGED X TO {x}')

func(x)
```

```python
x = 50

def func():

    global x
    print(f'X is {x}')

    # LOCAL REASSIGNMENT ON A GLOBAL VARIABLE!
    x = 'NEW VALUE'

    print(f'I JUST LOCALLY CHANGED GLOBAL X TO {x}')

func()

print(x)
```

```python
x = 50

def func(x):

    print(f'X is {x}')

    # LOCAL REASSIGNMENT ON A GLOBAL VARIABLE!
    x = 'NEW VALUE'

    print(f'I JUST LOCALLY CHANGED GLOBAL X TO {x}')

    return x

x = func(x)

print(x)
```
