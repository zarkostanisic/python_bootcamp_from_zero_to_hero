```python
def say_hello():
    print("hello")
    print("are")
    print('you')

say_hello()
say_hello
```

```python
def say_hello(name = 'Default'):
    print(f'Hello {name}')

say_hello('Zarko')
```

```python
def add_num(num1, num2):
    return num1 + num2

print(add_num(1, 2))

result = add_num(1, 2)

result
```

```python
def print_result(a, b):
    print(a + b)

def return_result(a, b):
    return a + b

result = print_result(1, 2)
print(type(result))

result = return_result(1, 2)
print(result)
print(type(result))
```

```python
def myfunc(a, b):
    print(a+b)

    return a + b

result = myfunc(1, 2)

result
```

```python
def sum_numbers(num1, num2):
    return num1 + num2

sum_numbers(1, 2)
sum_numbers('1', '2')
```
