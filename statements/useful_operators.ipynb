Useful operators

Range

```python
mylist = []

for num in range(10):
    print(num)
```

```python
for num in range(0, 10):
    print(num)
```

```python
for num in range(0, 10, 2):
    print(num)
```

```python
print(list(range(0, 11, 2)))
```

```python
index_count = 0

for letter in 'abcde':
    print('At index {} the letter is {}'.format(index_count, letter))
    index_count +=1
```

```python
word = 'abcde'

index_count = 0
for letter in word:
    print(word[index_count])
    index_count += 1

# tuple
for (index, letter) in enumerate(word):
    print(index)
    print(letter)
    print('\n')
```

Zip

```python
mylist1 = [1, 2, 3]
mylist2 = ["a", "b", "c"]

for item in zip(mylist1, mylist2):
    print(item)
```

```python
mylist1 = [1, 2, 3, 4, 5, 6]
mylist2 = ["a", "b", "c"]
mylist3 = [100, 200, 300]

for item in zip(mylist1, mylist2, mylist3):
    print(item)
```

```python
mylist1 = [1, 2, 3, 4, 5, 6]
mylist2 = ["a", "b", "c"]
mylist3 = [100, 200, 300]

for a, b, c in zip(mylist1, mylist2, mylist3):
    print(b)
```

```python
mylist1 = [1, 2, 3]
mylist2 = ["a", "b", "c"]

print(list(zip(mylist1, mylist2)))
```

```python
print('x' in [1, 2, 3])
```

```python
print('x' in ['x', 'y', 'z'])
```

```python
print('a' in 'a world')
```

```python
print('mykey' in {'mykey': 345})
```

Min, Max

```python
d = {'mykey': 345}

print(345 in d.values())
```

```python
mylist = [10, 20, 30, 40, 100]

print(min(mylist))
print(max(mylist))
```

Import shuffle, randint

```python
from random import shuffle

mylist = [10, 20, 30, 40, 100]

shuffle(mylist)

print(mylist) 
```

```python
from random import randint

mynum = randint(0, 100)
print(mynum)
```
