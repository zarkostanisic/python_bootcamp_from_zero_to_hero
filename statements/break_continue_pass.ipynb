Continue

```python
mystring = 'Sammy'

for letter in mystring:
    if letter == 'a':
        continue
    print(letter)
```

Break

```python
mystring = 'Sammy'

for letter in mystring:
    if letter == 'a':
        break
    print(letter)
```
