List comprehensions

```python
mystring = 'hello'

mylist = []

for letter in mystring:
    mylist.append(letter)

print(mylist)
```

```python
mystring = 'hello'

mylist = [letter for letter in mystring]

print(mylist)
```

```python
mylist = [x for x in 'word']

print(mylist)
```

```python
mylist = [num for num in range(0, 11)]

print(mylist)
```

```python
mylist = [num ** 2 for num in range(0, 11)]

print(mylist)
```

```python
mylist = [num for num in range(0, 11) if num % 2 == 0]

print(mylist)
```

```python
celsius = [0, 10, 30, 34.5]
fahrenheit = [((9 / 5) * temp + 32) for temp in celsius]

print(fahrenheit)
```

```python
celsius = [0, 10, 30, 34.5]
fahrenheit = []

for temp in celsius:
    fahrenheit.append((9 / 5) * temp + 32)

print(fahrenheit)
```

```python
results = [x if x % 2  == 0 else 'ODD' for x in range(0, 11)]

print(results)
```

```python
mylist = []

for x in [2, 4, 6]:
    for y in [10, 200, 300]:
        mylist.append(x * y)

print(mylist)

mylist = [x * y for x in[2,4,6] for y in [10, 200, 300]]

print(mylist)
```
