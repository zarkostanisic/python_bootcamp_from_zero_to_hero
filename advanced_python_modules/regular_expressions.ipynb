Regular expressions

```python
text = "The agent's phone number is 408-555-1234. Call soon!"
print('phone' in text)
```

```python
import re

text = "The agent's phone number is 408-555-1234. Call soon!"

pattern = 'phone'
print(re.search(pattern, text))

pattern = 'NOT IN TEXT'
print(re.search(pattern, text))

pattern = 'phone'
match = re.search(pattern, text)
print(match)
print(match.span())
print(match.start())
print(match.end())

text = 'my phone once, my phone twice'
match = re.search('phone', text)
print(match)

matches = re.findall('phone', text)
print(matches)
print(len(matches))

for match in re.finditer('phone', text):
    print(match.span())
    print(match.group())
```

Build Patterns

```python
text = 'My phone number is 408-555-7777-23'

phone = re.search(r'408-555-1234', text)
print(phone)
phone = re.search(r'\d\d\d-\d\d\d-\d\d\d\d', text)
print(phone)
#print(phone.group())

# + one or more times
# * zero or more times
# ? once or none
# {3} exactly 3 times
# {2, 4} 2 to 4 times
# {3,} 3 or more
phone = re.search(r'\d{3}+\d{3}-\d{4}?', text)
print(phone)
phone = re.search(r'\d{3}-+\d{3,3}-*\d{4,}(-\d{2})?', text)
print(phone)
```

```python
text = 'My phone number is 408-555-7777-23'

phone_pattern = re.compile(r'(\d{3})-(\d{3})-(\d{4})')
print(phone_pattern)

results = re.search(phone_pattern, text)
print(results.group(1))
print(results.group(2))
print(results.group(3))
```

Wildcard operator

```python
print(re.search(r'cat|dog', 'The dog is here'))

print(re.findall(r'.at', 'The cat in the hat sat there went splat'))
print(re.findall(r'...at', 'The cat in the hat sat there went splat'))
print(re.findall(r'^\d', '1 is a number'))
print(re.findall(r'^\d', 'The 2 is a number'))
print(re.findall(r'\d$', 'The number is 2'))
```

```python
phrase = 'there are 3 numbers 34 inside 5 this sentence'
pattern = r'[^\d]+'
print(re.findall(pattern, phrase))
```

Clean string

```python
test_phrase = 'This is a string! But it has punctuation. How can we remove it?'
clean = re.findall(r'[^!.? ]+', test_phrase)
' '.join(clean)
```

```python
text = 'Only find the hypen-words in this sentence. But you do not know how long-ish they are'

print(re.findall(r'[\w]+-[\w]+', text))

```
