Date time

```python
import datetime

mytime = datetime.time(13,20,1,2)
print(mytime.minute)
print(mytime.hour)
print(mytime)
print(mytime.microsecond)
print(type(mytime))
```

```python
today = datetime.date.today()
print(today)

print(today.year)
print(today.month)
print(today.day)
print(today.ctime())
```

```python
from datetime import datetime

mydatetime = datetime(2021,10,3,14)
print(mydatetime)

mydatetime = mydatetime.replace(year=2020)
print(mydatetime)
```

```python
# DATE
from datetime import date
date1 = date(2021,11,3)
date2 = date(2020,11,3)

result = date1 - date2
print(result)
print(type(result))
print(result.days)
```

```python
datetime1 = datetime(2021,11,3,22,0)
datetime2 = datetime(2020,11,3,12,0)

print(datetime1 - datetime2)
print(36000/60/60)

result = datetime1 - datetime2

print(result)
print(result.total_seconds())
```
