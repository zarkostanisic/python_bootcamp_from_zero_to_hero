OS Walk

```python
import os

os.getcwd()

file_path = os.getcwd()

for folder, sub_folders, files in os.walk(file_path):
    print(f"Currently looking ad {folder}")
    print("\n")

    print("The subfolders are: ")

    for sub_fold in sub_folders:
        print(f"\tSubfolder: {sub_fold}")

    print("\n")
    print("The files are: ")

    for f in files:
        print(f"\t File: {f}")
```
